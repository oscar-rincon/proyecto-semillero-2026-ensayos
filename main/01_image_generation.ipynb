# Binary Image Generation

```python
# Imports and reproducible configuration
import numpy as np

from utils.image_generation import (
    count_unique_images,
    generate_random_images,
    load_image_dataset,
    plot_image,
    save_image_dataset,
)

seed = 42
rng = np.random.default_rng(seed)

print(f"Seed used: {seed}")
```

## Dataset Creation and Saving

```python
# Generate a small sample while developing the project.
total_images = 100
image_size = (64, 64)

# Vectorized generation avoids creating thousands of small arrays.
images = generate_random_images(
    count=total_images,
    size=image_size,
    rng=rng,
)

# Check that there are no duplicate images.
unique_images = count_unique_images(images)
if unique_images != total_images:
    raise ValueError(f"Found {total_images - unique_images} duplicate images.")

print(f"Unique images: {unique_images}/{total_images}")

# Save the complete dataset and its metadata in one compressed file.
output_path = save_image_dataset(
    images=images,
    output_path="data/binary_image_dataset.npz",
    seed=seed,
)

print(f"Image size: {images.shape[1:]}")
print(f"File: data/binary_image_dataset.npz")

# Display one image as a visual check.
plot_image(images[0], title="Example binary image")
```

## Results and Verifications

```python
# Load and display the first image from the saved dataset.
loaded_images, loaded_seed, loaded_image_size = load_image_dataset(output_path)
loaded_image = loaded_images[0]

print(f"Loaded image shape: {loaded_image.shape}")
print(f"Loaded seed: {loaded_seed}")
print(f"Loaded image size: {loaded_image_size}")

plot_image(loaded_image, title="Loaded image")
```
